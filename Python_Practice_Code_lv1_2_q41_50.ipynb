# **43. 예산(250104)**

```python
def solution(d, budget):
    d.sort()  # 요청 금액을 오름차순으로 정렬하여 작은 금액부터 처리
    while budget < sum(d):
        # 현재 배열 d의 모든 요청 금액 합계(sum(d))가 예산(budget)을 초과하면:
        d.pop()
        # 가장 큰 금액(배열의 마지막 요소)을 제거하여 초과 금액을 줄임
    return len(d)
    # 남아 있는 요청 금액의 개수(d의 길이)를 반환 (지원 가능한 부서 수)
```

# **45. 삼총사(250107)**

# 주석 설명:
# 1. 세 개의 중첩 반복문을 사용하여 서로 다른 학생의 번호 조합을 순회합니다.
# 2. 첫 번째 반복문은 첫 번째 학생의 인덱스를 선택합니다.
# 3. 두 번째 반복문은 첫 번째 학생 이후의 학생 중 두 번째 학생을 선택합니다.
# 4. 세 번째 반복문은 두 번째 학생 이후의 학생 중 세 번째 학생을 선택합니다.
# 5. 각 조합의 합이 0이면 count를 증가시킵니다.
# 6. 최종적으로 count를 반환하여 삼총사의 수를 출력합니다.

```python
def solution(number):
    count = 0  # 삼총사의 수를 저장할 변수

    for x in range(len(number)):  # 첫 번째 학생 선택
        for y in range(x + 1, len(number)):  # 두 번째 학생 선택
            for z in range(y + 1, len(number)):  # 세 번째 학생 선택
                # 세 학생의 번호 합이 0이면 카운트 증가
                if number[x] + number[y] + number[z] == 0:
                    count += 1

    return count  # 삼총사의 개수 반환

# 주석 설명:
# 1. 세 개의 중첩 반복문을 사용하여 서로 다른 학생의 번호 조합을 순회합니다.
# 2. 첫 번째 반복문은 첫 번째 학생의 인덱스를 선택합니다.
# 3. 두 번째 반복문은 첫 번째 학생 이후의 학생 중 두 번째 학생을 선택합니다.
# 4. 세 번째 반복문은 두 번째 학생 이후의 학생 중 세 번째 학생을 선택합니다.
# 5. 각 조합의 합이 0이면 count를 증가시킵니다.
# 6. 최종적으로 count를 반환하여 삼총사의 수를 출력합니다.
```

# 주석 설명:
# 1. `itertools.combinations`를 사용해 학생 번호 리스트에서 3명을 뽑는 모든 조합을 생성합니다.
# 2. 각 조합의 합을 계산하여 0인지 확인합니다.
# 3. 조건을 만족하면 count를 증가시킵니다.
# 4. 반복문이 끝나면 최종적으로 count를 반환합니다.

```python
from itertools import combinations

def solution(number):
    count = 0  # 삼총사의 수를 저장할 변수

    # itertools.combinations로 학생 번호의 모든 3명 조합 생성
    for trio in combinations(number, 3):
        if sum(trio) == 0:  # 조합의 합이 0이면 카운트 증가
            count += 1

    return count  # 삼총사의 개수 반환

# 주석 설명:
# 1. `itertools.combinations`를 사용해 학생 번호 리스트에서 3명을 뽑는 모든 조합을 생성합니다.
# 2. 각 조합의 합을 계산하여 0인지 확인합니다.
# 3. 조건을 만족하면 count를 증가시킵니다.
# 4. 반복문이 끝나면 최종적으로 count를 반환합니다.
```

```python
def solution(s):
    stack = []  # 빈 스택을 초기화합니다.

    for char in s:  # 문자열의 문자를 하나씩 확인합니다.
        # 스택이 비어 있지 않고, 최상단 문자와 현재 문자가 같으면 제거
        if stack and stack[-1] == char:  # 스택 최상단 문자와 현재 문자가 같다면
            stack.pop()  # 최상단 문자를 제거합니다.
        else:
            stack.append(char)  # 그렇지 않으면 현재 문자를 스택에 추가합니다.

    return 1 if not stack else 0  # 스택이 비어 있으면 성공(1), 아니면 실패(0)
```

# **49. 최소직사각형(250113)** - 최종 오답

```python
def solution(sizes):
    # 각 명함의 가로(w)와 세로(h) 중 큰 값을 가로로 설정하고 작은 값을 세로로 설정
    # 긴 쪽을 항상 가로로 배치해 모든 명함이 수납될 수 있는 방향으로 고정
    rotated_sizes = [(max(w, h), min(w, h)) for w, h in sizes]

    # 모든 명함의 최대 가로 길이를 구함
    max_width = max(w for w, h in rotated_sizes)
    # 모든 명함의 최대 세로 길이를 구함
    max_height = max(h for w, h in rotated_sizes)

    # 최대 가로 길이와 최대 세로 길이를 곱하여 지갑의 최소 크기 계산
    return max_width * max_height

# 함수 동작 원리
# 1. sizes: 입력값으로 명함들의 가로(w)와 세로(h)가 담긴 2차원 배열입니다.
# 2. rotated_sizes: 각 명함의 긴 쪽(max(w, h))을 가로, 짧은 쪽(min(w, h))을 세로로 고정하여 새로운 리스트를 생성합니다.
# 3. max_width: 모든 명함의 가로 길이 중 가장 큰 값을 계산합니다.
# 4. max_height: 모든 명함의 세로 길이 중 가장 큰 값을 계산합니다.
# 5. max_width * max_height: 최대 가로와 최대 세로를 곱하여 지갑의 최소 크기를 반환합니다.

# 예제 입력과 출력 확인
print(solution([[60, 50], [30, 70], [60, 30], [80, 40]]))  # 4000
print(solution([[10, 7], [12, 3], [8, 15], [14, 7], [5, 15]]))  # 120
print(solution([[14, 4], [19, 6], [6, 16], [18, 7], [7, 11]]))  # 133
```

# 함수 동작 원리

1. max(w, h)와 min(w, h):

 - 각각 두 숫자 중 큰 값과 작은 값을 반환합니다.
 - 예: max(60, 50)은 60, min(60, 50)은 50을 반환합니다.

2. 리스트 내포 [(max(w, h), min(w, h)) for w, h in sizes]:

 - 입력된 sizes의 각 요소 [w, h]에서 긴 변과 짧은 변을 분리하여 튜플로 만듭니다.
 - 결과는 [(큰 변, 작은 변), ...] 형태의 리스트가 됩니다.

3. max(w for w, h in rotated_sizes):

 - 리스트 rotated_sizes에서 각 튜플의 첫 번째 값(긴 변)을 추출하여 그 중 가장 큰 값을 반환합니다.

4. 최종 계산:

 - 가장 큰 가로(max_width)와 세로(max_height) 값을 곱하여 최소 지갑 크기를 구합니다.

 - max([x[0] for x in sizes]) = 50 (짧은 쪽의 최대값)

```python
def solution(sizes):
    row = 0
    col = 0
    for a, b in sizes:
        if a < b:
            a, b = b, a
        row = max(row, a)
        col = max(col, b)
    return row * col
```

# 설명:

1. if a < b:

 - 각 명함의 가로(a), 세로(b) 중 큰 값(a)과 작은 값(b)을 구분합니다.
 - 필요 시 a와 b를 바꿔, a가 항상 큰 값을 가리키도록 합니다.

2. row와 col 업데이트:

 - row는 가로 길이의 최대값을 저장합니다.
 - col은 세로 길이의 최대값을 저장합니다.
 - 모든 명함을 순회하며 긴 쪽은 row, 짧은 쪽은 col로 업데이트합니다.

3. 곱하기:

 - 최종적으로 row * col을 계산하여 지갑의 최소 크기를 반환합니다.

```python
def solution(s, n):
    result = []  # 결과를 저장할 리스트
    for char in s:  # 문자열의 각 문자 순회
        if char.isupper():  # 대문자인 경우
            # 'A'를 기준으로 n만큼 밀고, 26으로 순환 처리
            result.append(chr((ord(char) - ord('A') + n) % 26 + ord('A')))
        elif char.islower():  # 소문자인 경우
            # 'a'를 기준으로 n만큼 밀고, 26으로 순환 처리
            result.append(chr((ord(char) - ord('a') + n) % 26 + ord('a')))
        else:
            result.append(char)  # 공백이나 다른 문자는 그대로 추가
    return ''.join(result)  # 리스트를 문자열로 합쳐서 반환
```

```python
def solution(s, n):
    answer = ''  # 최종 결과를 저장할 빈 문자열
    for i in s:  # 문자열의 각 문자를 순회
        if i:  # 문자가 비어있지 않은 경우 처리
            if i >= 'A' and i <= 'Z':  # 대문자인 경우
                answer += chr((ord(i) - ord('A') + n) % 26 + ord('A'))
                # 대문자 범위 내에서 시저 암호를 적용
            elif i >= 'a' and i <= 'z':  # 소문자인 경우
                answer += chr((ord(i) - ord('a') + n) % 26 + ord('a'))
                # 소문자 범위 내에서 시저 암호를 적용
            else:
                answer += ' '  # 공백은 그대로 추가
    return answer  # 최종 결과 반환
```

 - ord()는 문자를 정수로 변환합니다.
 - 이 정수는 해당 문자의 ASCII 코드(혹은 Unicode 코드포인트) 값을 나타냅니다.

예제 :
 - print(ord('A'))  # 출력: 65 ('A'의 ASCII 코드 값)
 - print(ord('a'))  # 출력: 97 ('a'의 ASCII 코드 값)

 - print(ord('z'))  # 출력: 122

동작 원리:
 - 컴퓨터는 문자(예: 'A', 'B', 'a')를 직접 저장하지 않고 숫자로 변환해 저장합니다.
 - ASCII는 영어 알파벳과 숫자, 일부 특수 문자를 표현하는 128개의 고유 코드 값입니다.
 -- 예: 'A' = 65, 'B' = 66, 'a' = 97, 'b' = 98
 - ord()는 특정 문자의 이 코드 값을 반환합니다.
