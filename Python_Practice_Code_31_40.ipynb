# **31. 행렬의 덧셈(241211)**

```python
def sumMatrix(A,B):
    answer = [[c + d for c, d in zip(a,b)] for a, b in zip(A,B)]
    return answer
```

설명:

zip(A, B): 두 행렬의 각 행을 묶어서 차례로 반환합니다. a와 b는 각각 A와 B의 같은 행에 해당합니다.

내부 zip(a, b): 각 행 a와 b를 다시 묶어 두 행의 같은 열 요소인 c와 d를 차례로 가져옵니다.

리스트 컴프리헨션: c + d를 계산하면서 새로운 행을 만듭니다.

최종 결과: 행렬 덧셈 결과가 완성됩니다.

```python
def sumMatrix(A,B):
    return [list(map(sum, zip(*x))) for x in zip(A, B)]
```

1) zip(*x) 설명

zip은 여러 개의 이터러블(리스트, 튜플 등)을 묶어서 순회하는 함수입니다.

*x는 언패킹 연산자입니다. *는 리스트나 튜플의 각 원소를 꺼내는 역할을 합니다.

예제와 함께 설명:

입력 값: x = ([1, 2], [3, 4])

여기서 zip(*x)는 *를 사용해 리스트 x의 원소를 각각 꺼냅니다.

즉, zip(*x)는 zip([1, 2], [3, 4])와 같아집니다.

zip([1, 2], [3, 4])는:

첫 번째 요소끼리 묶어서 (1, 3)

두 번째 요소끼리 묶어서 (2, 4)

결과: zip(*x)는 [(1, 3), (2, 4)]를 반환합니다.

핵심: zip(*x)는 행의 각 요소를 열 단위로 묶어주는 역할을 합니다.

2) map(sum, zip(*x)) 설명

map(function, iterable): map은 이터러블의 각 요소에 function을 적용해 결과를 반환합니다.

sum 함수: sum은 여러 숫자의 합을 구하는 함수입니다.

예제 설명:

zip(*x)의 결과가 [(1, 3), (2, 4)]라고 가정하면, map(sum, zip(*x))는 다음과 같은 동작을 합니다:

sum((1, 3)) → 1 + 3 = 4

sum((2, 4)) → 2 + 4 = 6

결과: map(sum, zip(*x))는 [4, 6]이라는 이터레이터를 반환합니다.

```python
def solution(arr1, arr2):
    return [[arr1[i][j] + arr2[i][j] for j in range(len(arr1[0]))] for i in range(len(arr1))]
```

코드 설명:

바깥 리스트 컴프리헨션: for i in range(len(arr1))를 사용해 행의 인덱스를 순회합니다.

안쪽 리스트 컴프리헨션: for j in range(len(arr1[0]))를 통해 각 행의 열 인덱스를 순회하면서 arr1[i][j] + arr2[i][j]를 계산합니다.

결과: 두 행렬의 각 행과 열의 같은 위치 값을 더해 최종 결과 행렬을 만듭니다.

# **32. 최댓값과 최솟값(241212)**

```python
def solution(s):
    numbers = list(map(int, s.split()))  # 문자열을 정수로 변환
    return str(min(numbers)) + ' ' + str(max(numbers))  # 최솟값과 최댓값 반환

# 문자열을 정수로 바꿔준 후 리스트업 -> 이후 최솟값 최댓값 뽑기 (문자열인 경우 정렬하면 음수와 양수 구분을 할수가 없음)
```

# **33. 직사각형 별찍기(241213)**

# 1) 줄바꿈을 기준으로 문자열 합치기(문자열에 줄바꿈 기호를 넣고 리스트를 join하면 리스트가 줄바꿈 기호를 기준으로 나뉨)

```python
def solution(a, b):
    result = []  # 결과를 담을 리스트 생성
    for i in range(b):  # b번 반복
        result.append('*' * a)  # 각 줄에 a개의 '*'를 추가
    return '\n'.join(result)  # 줄바꿈을 기준으로 문자열 합치기(문자열에 줄바꿈 기호를 넣고 리스트를 join하면 리스트가 줄바꿈 기호를 기준으로 나뉨)

# 이 경우는 출력값이 아닌 반환값으로 받았고 입력값이 없을경우이기 때문에 아래 코드를 활용
```

# 2) 기존 값이 주어지는경우 입력을 받는것부터 코딩

```python
a, b = map(int, input().split())  # 입력 받기
for i in range(b):  # b번 반복 (세로)
    print('*' * a)  # 별 a개 출력 (가로)
```
