 21. 없는 숫자 더하기(241118)

```python
# 문제 설명
# 0부터 9까지의 숫자 중 일부가 들어있는 정수 배열 numbers가 매개변수로 주어집니다.
# numbers에서 찾을 수 없는 0부터 9까지의 숫자를 모두 찾아 더한 수를
# return 하도록 solution 함수를 완성해주세요.

# 제한사항
# 1 ≤ numbers의 길이 ≤ 9
# 0 ≤ numbers의 모든 원소 ≤ 9
# numbers의 모든 원소는 서로 다릅니다.

# 입출력 예
# numbers	result
# [1,2,3,4,6,7,8,0]	14
# [5,8,4,0,6,7,9]	6

# 입출력 예 설명

# 입출력 예 #1
# 5, 9가 numbers에 없으므로, 5 + 9 = 14를 return 해야 합니다.

# 입출력 예 #2
# 1, 2, 3이 numbers에 없으므로, 1 + 2 + 3 = 6을 return 해야 합니다.
```

```python
def solution(numbers):
    a = sum(range(10))
    b = sum(int(x) for x in numbers)
    return a-b
```

 22. 제일 작은 수 제거하기(241120)

```python
# 문제 설명
# 정수를 저장한 배열, arr 에서 가장 작은 수를 제거한 배열을 리턴하는 함수,
# solution을 완성해주세요.
# 단, 리턴하려는 배열이 빈 배열인 경우엔 배열에 -1을 채워 리턴하세요.
# 예를들어 arr이 [4,3,2,1]인 경우는 [4,3,2]를 리턴 하고, [10]면 [-1]을 리턴 합니다.

# 제한 조건
# arr은 길이 1 이상인 배열입니다.
# 인덱스 i, j에 대해 i ≠ j이면 arr[i] ≠ arr[j] 입니다.

# 입출력 예
# arr	return
# [4,3,2,1]	[4,3,2]
# [10]	[-1]
```

1) 내가 쓴 답

```python
def solution(arr):
    arr.pop(arr.index(min(arr)))
    return arr if arr else [-1]
```

```python
solution([10])
```

```python
solution([4,3,2,1])
```

2) gpt가 쓴 답

```python
def solution(arr):
    # 배열의 길이가 1이면 [-1] 반환
    if len(arr) == 1:
        return [-1]

    # 배열에서 가장 작은 값 제거
    min_value = min(arr)  # 배열에서 가장 작은 값 찾기
    arr.remove(min_value)  # 가장 작은 값 제거
    return arr
```

3) 다른 사람들이 쓴 답

```python
def rm_small(mylist):
    # 함수를 완성하세요
    return [i for i in mylist if i > min(mylist)]


# 아래는 테스트로 출력해 보기 위한 코드입니다.
my_list = [4, 3, 2, 1]
print("결과 {} ".format(rm_small(my_list)))
```

23. 핸드폰 번호 가리기(241125)

```python
# 문제 설명
# 프로그래머스 모바일은 개인정보 보호를 위해
# 고지서를 보낼 때 고객들의 전화번호의 일부를 가립니다.
# 전화번호가 문자열 phone_number로 주어졌을 때,
# 전화번호의 뒷 4자리를 제외한 나머지 숫자를 전부 *으로 가린 문자열을 리턴하는 함수,
# solution을 완성해주세요.

# 제한 조건
# phone_number는 길이 4 이상, 20이하인 문자열입니다.
# 입출력 예
# phone_number	return
# "01033334444"	"*******4444"
# "027778888"	"*****8888"
```

1) 내가 쓴 답

```python
def solution(phone_number):
    return ''.join('*' for _ in range(len(phone_number) - 4)) + phone_number[-4:]
```

2) gpt가 쓴 답

```python
def solution(phone_number):
    return '*' * (len(phone_number) - 4) + phone_number[-4:]
```

```python
def solution(phone_num):
    answer = ''
    # 마지막 4자리를 제외한 앞 부분을 '*'로 대체
    for i in range(len(phone_num) - 4):
        answer += '*'
    # 마지막 4자리를 그대로 붙이기
    answer += phone_num[-4:]
    return answer
```

 24. 내적(241126)

```python
# 문제 설명
# 길이가 같은 두 1차원 정수 배열 a, b가 매개변수로 주어집니다.
# a와 b의 내적을 return 하도록 solution 함수를 완성해주세요.

# 이때, a와 b의 내적은 a[0]*b[0] + a[1]*b[1] + ... + a[n-1]*b[n-1] 입니다.
# (n은 a, b의 길이)

# 제한사항
# a, b의 길이는 1 이상 1,000 이하입니다.
# a, b의 모든 수는 -1,000 이상 1,000 이하입니다.

# 입출력 예
# a	b	result
# [1,2,3,4]	[-3,-1,0,2]	3
# [-1,0,1]	[1,0,-1]	-2

# 입출력 예 설명

# 입출력 예 #1
# a와 b의 내적은 1*(-3) + 2*(-1) + 3*0 + 4*2 = 3 입니다.

# 입출력 예 #2
# a와 b의 내적은 (-1)*1 + 0*0 + 1*(-1) = -2 입니다.
```

1) 내가 쓴 답

```python
def solution(a, b):
    answer = 0
    for i in range(len(a)):
        answer += a[i]*b[i]
    return answer
```

```python
a=[1,2,3,4]
```

```python
b=[-3,-1,0,2]
```

```python
solution(a, b)
```

2) 다른 사람들이 쓴 답

```python
def solution(a, b):

    return sum([x*y for x, y in zip(a,b)])
```

3) gpt가 개선해 준 답

```python
def solution(a, b):
    answer = 0
    for a, b in zip(a, b):
        answer += a*b
    return answer
```

```python
a=[1,2,3,4]
```

```python
b=[-3,-1,0,2]
```

```python
solution(a, b)
```

25. 가운데 글자 가져오기(241203)

```python
# 문제 설명
# 단어 s의 가운데 글자를 반환하는 함수, solution을 만들어 보세요.
# 단어의 길이가 짝수라면 가운데 두글자를 반환하면 됩니다.

# 재한사항
# s는 길이가 1 이상, 100이하인 스트링입니다.

# 입출력 예
# s	return
# "abcde"	"c"
# "qwer"	"we"
```

1) 내가 쓴 답

```python
def solution(s):
    if len(s) % 2 == 0:
      answer = s[(len(s)//2)-1]+s[len(s)//2]
    else:
      answer = s[(len(s)//2)]
    return answer
```

```python
s = "abcde"
```

26. 수박수박수박수박수박수?(241204)

```python
# 문제 설명
# 길이가 n이고, "수박수박수박수...."와 같은
# 패턴을 유지하는 문자열을 리턴하는 함수, solution을 완성하세요.
# 예를들어 n이 4이면 "수박수박"을 리턴하고 3이라면 "수박수"를 리턴하면 됩니다.

# 제한 조건
# n은 길이 10,000이하인 자연수입니다.

# 입출력 예
# n	return
# 3	"수박수"
# 4	"수박수박"
```

```python
def solution(n):
    wm = '수박'
    if n % 2 == 0:
      answer = wm*(n//2)
    else:
      answer = wm*(n//2)+wm[0]
    return answer
```

27. 약수의 개수와 덧셈(241205) - 오답

```python
# 문제 설명
# 두 정수 left와 right가 매개변수로 주어집니다.
# left부터 right까지의 모든 수들 중에서, 약수의 개수가 짝수인 수는 더하고,
# 약수의 개수가 홀수인 수는 뺀 수를 return 하도록 solution 함수를 완성해주세요.

# 제한사항
# 1 ≤ left ≤ right ≤ 1,000

# 입출력 예
# left	right	result
# 13	17	43
# 24	27	52

# 입출력 예 설명

# 입출력 예 #1
# 다음 표는 13부터 17까지의 수들의 약수를 모두 나타낸 것입니다.

# 수	약수	약수의 개수
# 13	1, 13	2
# 14	1, 2, 7, 14	4
# 15	1, 3, 5, 15	4
# 16	1, 2, 4, 8, 16	5
# 17	1, 17	2

# 따라서, 13 + 14 + 15 - 16 + 17 = 43을 return 해야 합니다.

# 입출력 예 #2
# 다음 표는 24부터 27까지의 수들의 약수를 모두 나타낸 것입니다.

# 수	약수	약수의 개수
# 24	1, 2, 3, 4, 6, 8, 12, 24	8
# 25	1, 5, 25	3
# 26	1, 2, 13, 26	4
# 27	1, 3, 9, 27	4

# 따라서, 24 - 25 + 26 + 27 = 52를 return 해야 합니다.
```

1) 내가 쓴 답 - 틀림

```python
def solution(left, right):
    ls = left
    sum = 0
    for i in range(right-ls):
      if ls <= right:
        count = 0
        for j in range(ls+1):
          if j == 0:
            continue
          elif ls % j == 0:
            count += 1
            if j <= ls:
              if count % 2 == 0:
                sum += ls
              else:
                sum -= ls
            else:
              break
      ls += 1
    answer = sum
    return answer
```

```python
solution(13, 17)
```

1) -1 내가 쓴 답 개선

```python
def solution(left, right):
    total_sum = 0

    for num in range(left, right + 1):
        count = 0  # 약수 개수 초기화
        for i in range(1, num + 1):
            if num % i == 0:
                count += 1

        # 약수 개수에 따른 합산
        if count % 2 == 0:
            total_sum += num  # 약수의 개수가 짝수인 경우 더함
        else:
            total_sum -= num  # 약수의 개수가 홀수인 경우 뺌

    return total_sum
```

```python
solution(13, 17)
```

2) gpt가 쓴 답

```python
def solution(left, right):
    total_sum = 0

    for num in range(left, right + 1):
        # 제곱수인지 판별 (제곱수인 경우 약수 개수가 홀수)
        if int(num ** 0.5) == num ** 0.5:  # 정수로 나누어 떨어지면 제곱수
            total_sum -= num  # 약수의 개수가 홀수인 경우 뺌
        else:
            total_sum += num  # 약수의 개수가 짝수인 경우 더함

    return total_sum
```

```python
solution(13, 17)
```

2) - 1 gpt가 쓴 답

```python
import math

def solution(left, right):
    total_sum = 0

    for num in range(left, right + 1):
        # 제곱수 판별: num의 제곱근이 정수인지 확인
        if math.isqrt(num) ** 2 == num:  # 정수 제곱 여부 확인
            total_sum -= num  # 약수 개수가 홀수인 경우 (제곱수) 뺌
        else:
            total_sum += num  # 약수 개수가 짝수인 경우 더함

    return total_sum
```

```python
solution(13, 17)
```

28. 문자열 내림차순으로 배치하기(241209)

문제 설명

문자열 s에 나타나는 문자를 큰것부터 작은 순으로 정렬해 새로운 문자열을 리턴하는 함수, solution을 완성해주세요.

s는 영문 대소문자로만 구성되어 있으며, 대문자는 소문자보다 작은 것으로 간주합니다.

제한 사항

str은 길이 1 이상인 문자열입니다.

입출력 예

s	return

"Zbcdefg"	"gfedcbZ"

1) 내가 쓴 답

```python
def solution(s):
    return ''.join(sorted(s, reverse=True))
```
