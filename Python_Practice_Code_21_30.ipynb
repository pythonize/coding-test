22. 제일 작은 수 제거하기(241120)

```python
def solution(arr):
    arr.pop(arr.index(min(arr)))
    return arr if arr else [-1]


# min(리스트) -> 리스트중 최솟값(대상값)
# 리스트.index(대상값) -> 위치 숫자값
# 리스트.pop(위치 숫자값)
# return arr if arr else [-1] 있으면 가져오고 없으면 -1
```

```python
def solution(arr):
    # 배열의 길이가 1이면 [-1] 반환
    if len(arr) == 1:
        return [-1]

    # 배열에서 가장 작은 값 제거
    min_value = min(arr)  # 배열에서 가장 작은 값 찾기
    arr.remove(min_value)  # 가장 작은 값 제거
    return arr
```

```python
def rm_small(mylist):
    return [i for i in mylist if i > min(mylist)]
# 나머지값이 최솟값보다 클경우 차례대로 리스트 만들어라

# 아래는 포맷사용법
my_list = [4, 3, 2, 1]
print("결과 {} ".format(rm_small(my_list)))
```

27. 약수의 개수와 덧셈(241205)

```python
def solution(left, right):
    total_sum = 0

    for num in range(left, right + 1):    # 범위를 쉼표로 지정해주기
        count = 0  # 약수 개수 초기화
        for i in range(1, num + 1):
            if num % i == 0:
                count += 1

        # 약수 개수에 따른 합산
        if count % 2 == 0:
            total_sum += num  # 약수의 개수가 짝수인 경우 더함
        else:
            total_sum -= num  # 약수의 개수가 홀수인 경우 뺌

    return total_sum
```

```python
import math

def solution(left, right):
    total_sum = 0

    for num in range(left, right + 1):
        # 제곱수 판별: num의 제곱근이 정수인지 확인 -> 제곱근이 정수일경우(예를들어 9, 16, 25 등일 경우 약수개수가 홀수임)
        if math.isqrt(num) ** 2 == num:  # 정수 제곱 여부 확인
            total_sum -= num  # 약수 개수가 홀수인 경우 (제곱수) 뺌
        else:
            total_sum += num  # 약수 개수가 짝수인 경우 더함

    return total_sum
```
