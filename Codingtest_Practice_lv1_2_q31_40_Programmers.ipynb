# **31. 행렬의 덧셈(241211)**

문제 설명

행렬의 덧셈은 행과 열의 크기가 같은 두 행렬의 같은 행, 같은 열의 값을 서로 더한 결과가 됩니다.
2개의 행렬 arr1과 arr2를 입력받아, 행렬 덧셈의 결과를 반환하는 함수, solution을 완성해주세요.

제한 조건

행렬 arr1, arr2의 행과 열의 길이는 500을 넘지 않습니다.

입출력 예

arr1	arr2	return

[[1,2],[2,3]]	[[3,4],[5,6]]	[[4,6],[7,9]]

[[1],[2]]	[[3],[4]]	[[4],[6]]

1) 내가 쓴 답

```python
def solution(arr1, arr2):
    answer = []
    for i in range(len(arr1)):
      list = []
      for j in range(len(arr1[0])):
        list.append(arr1[i][j] + arr2[i][j])
        if j == len(arr1[0])-1:
          answer.append(list)
    return answer
```

```python
arr1 = [[1,2],[2,3]]
```

```python
arr2 = [[3,4],[5,6]]
```

```python
solution(arr1, arr2)
```

2) 다른 사람들이 쓴 답

```python
def sumMatrix(A,B):
    answer = [[c + d for c, d in zip(a,b)] for a, b in zip(A,B)]
    return answer

# 아래는 테스트로 출력해 보기 위한 코드입니다.
print(sumMatrix([[1,2], [2,3]], [[3,4],[5,6]]))
```

설명:

zip(A, B): 두 행렬의 각 행을 묶어서 차례로 반환합니다. a와 b는 각각 A와 B의 같은 행에 해당합니다.

내부 zip(a, b): 각 행 a와 b를 다시 묶어 두 행의 같은 열 요소인 c와 d를 차례로 가져옵니다.

리스트 컴프리헨션: c + d를 계산하면서 새로운 행을 만듭니다.

최종 결과: 행렬 덧셈 결과가 완성됩니다.

```python
def sumMatrix(A,B):
    return [list(map(sum, zip(*x))) for x in zip(A, B)]
```

설명:

zip(A, B): 두 행렬 A와 B를 행 단위로 묶어 반환합니다. 이 결과는 x에 해당합니다.

zip(*x): x는 두 행렬의 같은 행이 묶인 값입니다. * 연산자를 사용해 x를 열 단위로 묶습니다.

예를 들어 x = ([1, 2], [3, 4])일 때, zip(*x)는 (1, 3)과 (2, 4)를 반환합니다.

map(sum, ...): zip(*x)의 결과인 각 열 요소를 합산합니다.

list(): map의 결과를 리스트로 변환해 새로운 행을 만듭니다.

최종 결과: 리스트 컴프리헨션을 통해 각 행을 완성하고 이를 반환합니다.

안쪽 리스트 컴프리헨션: for j in range(len(arr1[0]))를 통해 각 행의 열 인덱스를 순회하면서 arr1[i][j] + arr2[i][j]를 계산합니다.

# **34. 최솟값 만들기(241216)**

문제 설명
길이가 같은 배열 A, B 두개가 있습니다. 각 배열은 자연수로 이루어져 있습니다.

배열 A, B에서 각각 한 개의 숫자를 뽑아 두 수를 곱합니다.

이러한 과정을 배열의 길이만큼 반복하며, 두 수를 곱한 값을 누적하여 더합니다.

이때 최종적으로 누적된 값이 최소가 되도록 만드는 것이 목표입니다.

 (단, 각 배열에서 k번째 숫자를 뽑았다면 다음에 k번째 숫자는 다시 뽑을 수 없습니다.)

예를 들어 A = [1, 4, 2] , B = [5, 4, 4] 라면

A에서 첫번째 숫자인 1, B에서 첫번째 숫자인 5를 뽑아 곱하여 더합니다.
(누적된 값 : 0 + 5(1x5) = 5)

A에서 두번째 숫자인 4, B에서 세번째 숫자인 4를 뽑아 곱하여 더합니다.
(누적된 값 : 5 + 16(4x4) = 21)

A에서 세번째 숫자인 2, B에서 두번째 숫자인 4를 뽑아 곱하여 더합니다.
(누적된 값 : 21 + 8(2x4) = 29)

즉, 이 경우가 최소가 되므로 29를 return 합니다.

배열 A, B가 주어질 때 최종적으로 누적된 최솟값을 return 하는 solution 함수를 완성해 주세요.

제한사항

배열 A, B의 크기 : 1,000 이하의 자연수

배열 A, B의 원소의 크기 : 1,000 이하의 자연수

입출력 예

A	B	answer

[1, 4, 2]	[5, 4, 4]	29

[1,2]	[3,4]	10

입출력 예 설명

입출력 예 #1

문제의 예시와 같습니다.

입출력 예 #2

A에서 첫번째 숫자인 1, B에서 두번째 숫자인 4를 뽑아 곱하여 더합니다.

 (누적된 값 : 4) 다음, A에서 두번째 숫자인 2, B에서 첫번째 숫자인 3을 뽑아 곱하여 더합니다.

 (누적된 값 : 4 + 6 = 10)
이 경우가 최소이므로 10을 return 합니다.

1) 내가 쓴 답

```python
def solution(A, B):
    answer = []
    a = sorted(A, reverse=True)
    b = sorted(B, reverse=False)
    for i in range(len(a)):
        answer.append(a[i]*b[i])
    return sum(answer)
```

```python
A = [1,2]
```

```python
B = [3,4]
```

```python
solution(A, B)
```

2) 다른 사람들이 쓴 답

```python
def getMinSum(A, B):
    return sum([a * b for a, b in zip(sorted(A), sorted(B, reverse=True))])
```

배열 arr에서 연속적으로 나타나는 숫자는 제거하고 남은 수들을 return 하는 solution 함수를 완성해 주세요.

```python
def solution(arr):
    answer = [arr[0]]  # 첫 번째 원소는 무조건 추가
    for i in range(1, len(arr)):
        if arr[i] != arr[i-1]:  # 현재 원소와 이전 원소가 다르면 추가
            answer.append(arr[i])
    return answer
```

# **36. 올바른 괄호(241219)**

문제 설명
괄호가 바르게 짝지어졌다는 것은 '(' 문자로 열렸으면

반드시 짝지어서 ')' 문자로 닫혀야 한다는 뜻입니다.

예를 들어

"()()" 또는 "(())()" 는 올바른 괄호입니다.

")()(" 또는 "(()(" 는 올바르지 않은 괄호입니다.

'(' 또는 ')' 로만 이루어진 문자열 s가 주어졌을 때, 문자열 s가 올바른 괄호이면 true를 return 하고,

올바르지 않은 괄호이면 false를 return 하는 solution 함수를 완성해 주세요.

제한사항

문자열 s의 길이 : 100,000 이하의 자연수

문자열 s는 '(' 또는 ')' 로만 이루어져 있습니다.

입출력 예

s	answer

"()()"	true

"(())()"	true

")()("	false

"(()("	false

입출력 예 설명

입출력 예 #1,2,3,4

문제의 예시와 같습니다.

1) 내가 쓴 답

```python
def solution(s):
    if s[0] == ')':
        return False
    else:
      count = 0
      for i in s:
        if i == '(':
          count += 1
        else:
          count -= 1
          if count == -1:
            return False
      if count == 0:
        return True
      else:
        return False
```

```python
s = "(())()"
```

```python
solution(s)
```

2) gpt가 개선해준 코드

개선 포인트

1. 불필요한 조건문 제거: if s[0] == ')' 부분은 count가 음수가 되는 순간을 확인하므로 필요 없습니다.

2. 코드 간결화: if count == 0: return True 대신 return count == 0으로 바로 반환하도록 개선했습니다.

3. 가독성 향상: else 블록을 줄이고, 직관적인 조건으로 코드를 정리했습니다.

```python
def solution(s):
    count = 0
    for char in s:
        if char == '(':
            count += 1
        else:
            count -= 1
            if count < 0:  # 음수가 되는 순간 닫는 괄호가 더 많으므로 False
                return False
    return count == 0  # 모든 문자를 돌았을 때 count가 0이면 True
```

3) gpt가 쓴 코드

```python
def solution(s):
    stack = []  # 스택을 초기화합니다. 여는 괄호 '('를 저장하기 위해 사용됩니다.

    for char in s:  # 문자열 s를 한 문자씩 순회합니다.
        if char == '(':
            # 여는 괄호 '('를 만나면 스택에 추가합니다.
            stack.append(char)
        elif stack:
            # 닫는 괄호 ')'를 만나면:
            # 스택에 여는 괄호 '('가 이미 있을 경우, 짝을 맞춰 하나를 제거합니다.
            stack.pop()
        else:
            # 스택이 비어있는데 닫는 괄호 ')'가 등장한 경우
            # 여는 괄호가 부족하므로 바로 False를 반환합니다.
            return False

    # 문자열 전체를 순회한 후:
    # 스택이 비어있으면 모든 괄호가 짝지어졌으므로 True를 반환합니다.
    # 스택에 값이 남아있다면 여는 괄호 '('가 남아있다는 의미이므로 False를 반환합니다.
    return not stack
```

# **37. 최대공약수와 최소공배수(241223)**

문제 설명
두 수를 입력받아 두 수의 최대공약수와 최소공배수를 반환하는 함수, solution을 완성해 보세요.

배열의 맨 앞에 최대공약수, 그다음 최소공배수를 넣어 반환하면 됩니다.

예를 들어 두 수 3, 12의 최대공약수는 3, 최소공배수는 12이므로 solution(3, 12)는 [3, 12]를 반환해야 합니다.

제한 사항

두 수는 1이상 1000000이하의 자연수입니다.

입출력 예

n	m	return

3	12	[3, 12]

2	5	[1, 10]

입출력 예 설명

입출력 예 #1

위의 설명과 같습니다.

입출력 예 #2

자연수 2와 5의 최대공약수는 1, 최소공배수는 10이므로 [1, 10]을 리턴해야 합니다.

1) 내가 쓴 답

```python
import math

def solution(n, m):
    answer = []
    a = math.gcd(n, m)
    b = (n*m)//math.gcd(n,m)
    answer.append(a)
    answer.append(b)
    return answer
```

```python
def solution(n, m):
    # 최대공약수(GCD)를 계산하는 함수 (유클리드 호제법 사용)
    def gcd(a, b):
        # b가 0이 될 때까지 반복
        while b:
            # a에 b를 대입하고, b에 a를 b로 나눈 나머지를 대입
            a, b = b, a % b
        # 반복문 종료 후, a에는 최대공약수가 저장되어 있음
        return a

    # 최소공배수(LCM)를 계산하는 함수
    def lcm(a, b):
        # 두 수의 곱을 최대공약수로 나눈 값이 최소공배수
        return a * b // gcd(a, b)

    # gcd 함수와 lcm 함수를 호출하여 결과 리스트로 반환
    return [gcd(n, m), lcm(n, m)]

# 테스트
print(solution(3, 12))  # [3, 12] -> 3과 12의 최대공약수는 3, 최소공배수는 12
print(solution(2, 5))   # [1, 10] -> 2와 5의 최대공약수는 1, 최소공배수는 10
```

```python
def solution(s):
    answer = ''
    ch = s.split()
    for i in range(len(ch)):
      if i > 0:
        answer += ' '
      for j in range(len(ch[i])):
        if j == 0:
          if ch[i][j].islower():
            answer += ch[i][j].upper()
          else:
            answer += ch[i][j]
        else:
          if ch[i][j].isupper():
            answer += ch[i][j].lower()
          else:
            answer += ch[i][j]
    return answer
```
