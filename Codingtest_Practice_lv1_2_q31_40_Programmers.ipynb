# **31. 행렬의 덧셈(241211)**

문제 설명

행렬의 덧셈은 행과 열의 크기가 같은 두 행렬의 같은 행, 같은 열의 값을 서로 더한 결과가 됩니다.
2개의 행렬 arr1과 arr2를 입력받아, 행렬 덧셈의 결과를 반환하는 함수, solution을 완성해주세요.

제한 조건

행렬 arr1, arr2의 행과 열의 길이는 500을 넘지 않습니다.

입출력 예

arr1	arr2	return

[[1,2],[2,3]]	[[3,4],[5,6]]	[[4,6],[7,9]]

[[1],[2]]	[[3],[4]]	[[4],[6]]

1) 내가 쓴 답

```python
def solution(arr1, arr2):
    answer = []
    for i in range(len(arr1)):
      list = []
      for j in range(len(arr1[0])):
        list.append(arr1[i][j] + arr2[i][j])
        if j == len(arr1[0])-1:
          answer.append(list)
    return answer
```

```python
arr1 = [[1,2],[2,3]]
```

```python
arr2 = [[3,4],[5,6]]
```

```python
solution(arr1, arr2)
```

2) 다른 사람들이 쓴 답

```python
def sumMatrix(A,B):
    answer = [[c + d for c, d in zip(a,b)] for a, b in zip(A,B)]
    return answer

# 아래는 테스트로 출력해 보기 위한 코드입니다.
print(sumMatrix([[1,2], [2,3]], [[3,4],[5,6]]))
```

설명:

zip(A, B): 두 행렬의 각 행을 묶어서 차례로 반환합니다. a와 b는 각각 A와 B의 같은 행에 해당합니다.

내부 zip(a, b): 각 행 a와 b를 다시 묶어 두 행의 같은 열 요소인 c와 d를 차례로 가져옵니다.

리스트 컴프리헨션: c + d를 계산하면서 새로운 행을 만듭니다.

최종 결과: 행렬 덧셈 결과가 완성됩니다.

```python
def sumMatrix(A,B):
    return [list(map(sum, zip(*x))) for x in zip(A, B)]
```

설명:

zip(A, B): 두 행렬 A와 B를 행 단위로 묶어 반환합니다. 이 결과는 x에 해당합니다.

zip(*x): x는 두 행렬의 같은 행이 묶인 값입니다. * 연산자를 사용해 x를 열 단위로 묶습니다.

예를 들어 x = ([1, 2], [3, 4])일 때, zip(*x)는 (1, 3)과 (2, 4)를 반환합니다.

map(sum, ...): zip(*x)의 결과인 각 열 요소를 합산합니다.

list(): map의 결과를 리스트로 변환해 새로운 행을 만듭니다.

최종 결과: 리스트 컴프리헨션을 통해 각 행을 완성하고 이를 반환합니다.

안쪽 리스트 컴프리헨션: for j in range(len(arr1[0]))를 통해 각 행의 열 인덱스를 순회하면서 arr1[i][j] + arr2[i][j]를 계산합니다.

# **32. 최댓값과 최솟값(241212)**(오답)

문제 설명

문자열 s에는 공백으로 구분된 숫자들이 저장되어 있습니다.
str에 나타나는 숫자 중 최소값과 최대값을 찾아 이를 "(최소값) (최대값)"형태의 문자열을 반환하는 함수, solution을 완성하세요.

예를들어 s가 "1 2 3 4"라면 "1 4"를 리턴하고, "-1 -2 -3 -4"라면 "-4 -1"을 리턴하면 됩니다.

제한 조건

s에는 둘 이상의 정수가 공백으로 구분되어 있습니다.

입출력 예

s	return

"1 2 3 4"	"1 4"

"-1 -2 -3 -4"	"-4 -1"

"-1 -1"	"-1 -1"

1) 내가 쓴 답(오답)

```python
def solution(s):
    return sorted(s.split())[0] + ' ' + sorted(s.split())[-1]
```

```python
s = "-1 -2 -3 -4"
```

```python
solution(s)
```

2) gpt가 가르쳐준 답

```python
def solution(s):
    numbers = list(map(int, s.split()))  # 문자열을 정수로 변환
    return str(min(numbers)) + ' ' + str(max(numbers))  # 최솟값과 최댓값 반환
```

```python
s = "-1 -2 -3 -4"
```

```python
solution(s)
```

```python
s = "1 2 3 4"
```

```python
solution(s)
```

33. 직사각형 별찍기(241213)

문제 설명

이 문제에는 표준 입력으로 두 개의 정수 n과 m이 주어집니다.

별(*) 문자를 이용해 가로의 길이가 n, 세로의 길이가 m인 직사각형 형태를 출력해보세요.

제한 조건

n과 m은 각각 1000 이하인 자연수입니다.

예시

입력

5 3

출력

'*****'

'*****'

'*****'

1) 내가 쓴 답

절반 답 ->

입력값이 주어지기 때문에 받을 함수가 있어야함

그리고 출력값을 받아야하기때문에 return 대신 print로 받아야함

```python
def solution(a, b):
    for i in range(b):
      print('*'*a)
```

```python
solution(5, 3)
```

2) gpt가 개선한 답

```python
def solution(a, b):
    result = []  # 결과를 담을 리스트 생성
    for i in range(b):  # b번 반복
        result.append('*' * a)  # 각 줄에 a개의 '*'를 추가
    return '\n'.join(result)  # 줄바꿈을 기준으로 문자열 합치기

# 이 경우는 출력값이 아닌 반환값으로 받았고 입력값이 없을경우이기 때문에 아래 코드를 활용
```

3) 입력값이 있을경우 입력을 받는 함수가 필요, 출력값을 내는 코드가 필요

```python
a, b = map(int, input().split())  # 입력 받기
for i in range(b):  # b번 반복 (세로)
    print('*' * a)  # 별 a개 출력 (가로)
```
