17. 나누어 떨어지는 숫자 배열(241111)

```python
def solution(arr, divisor):
    answer = [x for x in arr if x % divisor == 0]  # divisor로 나누어 떨어지는 값만 리스트에 포함
    if not answer:  # 나누어 떨어지는 값이 없으면
        return [-1]
    return sorted(answer)  # 오름차순으로 정렬하여 반환
```

19. 콜라츠 추측(241113)

```python
def solution(num):
    num = int(num)  # num을 정수형으로 변환하여 안정성 확보

    if num == 1:
        return 0

    count = 0  # 반복 횟수 초기화

    while num != 1:
        if count >= 500:
            return -1
        num = num // 2 if num % 2 == 0 else num * 3 + 1
        count += 1

    return count
```
